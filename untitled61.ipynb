```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader, random_split
import matplotlib.pyplot as plt

print("📥 Descargando dataset de Flores (puede tardar 1 minuto)...")
# Data Augmentation (como querías)
transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomCrop(64, padding=4),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

# Descarga automática de las flores
full_dataset = torchvision.datasets.Flowers102(root='./data', split='train', download=True, transform=transform)
num_classes = 102

# Dividimos los datos
train_size = int(0.8 * len(full_dataset))
val_size = len(full_dataset) - train_size
train_dataset, val_dataset = random_split(full_dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)

# ══════════════════════════════════════════════════════════════════
# TU FUNCIÓN GENAL
# ══════════════════════════════════════════════════════════════════
class GenalShift(nn.Module):
    def __init__(self, theta_init: float = 1.0, beta_init: float = 0.0):
        super().__init__()
        self.theta = nn.Parameter(torch.tensor(theta_init, dtype=torch.float32))
        self.beta = nn.Parameter(torch.tensor(beta_init, dtype=torch.float32))
    def forward(self, x):
        k = torch.clamp(torch.nn.functional.softplus(self.theta), min=1e-6, max=2.0)
        return x * torch.sigmoid((x - self.beta) / k)

# ══════════════════════════════════════════════════════════════════
# EL MODELO (CNN)
# ══════════════════════════════════════════════════════════════════
class BotanicoCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
        self.act1 = GenalShift()
        self.pool = nn.MaxPool2d(2, 2)

        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.act2 = GenalShift()

        self.fc1 = nn.Linear(32 * 16 * 16, 128)
        self.act3 = GenalShift()
        self.drop = nn.Dropout(0.5) # <-- El escudo anti-memorización
        self.fc2 = nn.Linear(128, num_classes)

    def forward(self, x):
        x = self.pool(self.act1(self.conv1(x)))
        x = self.pool(self.act2(self.conv2(x)))
        x = x.view(-1, 32 * 16 * 16)
        x = self.drop(self.act3(self.fc1(x)))
        x = self.fc2(x)
        return x

# ══════════════════════════════════════════════════════════════════
# ENTRENAMIENTO
# ══════════════════════════════════════════════════════════════════
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
modelo = BotanicoCNN(num_classes).to(device)
criterio = nn.CrossEntropyLoss()
optimizer = optim.Adam(modelo.parameters(), lr=0.001)

print("🚀 Iniciando entrenamiento con GenalShift...")
epocas = 10
historial_loss = []

for epoch in range(epocas):
    modelo.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = modelo(images)
        loss = criterio(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    avg_loss = running_loss / len(train_loader)
    historial_loss.append(avg_loss)
    print(f"Época {epoch+1}/{epocas} | Pérdida: {avg_loss:.4f}")

# Validación
modelo.eval()
correct = 0
total = 0
with torch.no_grad():
    for images, labels in val_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = modelo(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

precision = 100 * correct / total
print(f"\n🏆 ¡Entrenamiento terminado!")
print(f"🎯 Precisión (Accuracy): {precision:.2f}%")

plt.figure(figsize=(8, 4))
plt.plot(historial_loss, marker='o', color='blue', linewidth=2)
plt.title("Aprendizaje de Flores (usando GenalShift)", fontweight='bold')
plt.xlabel("Épocas")
plt.ylabel("Pérdida (Loss)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader, random_split
import matplotlib.pyplot as plt

print("📥 Descargando dataset de Flores...")
# AHORA USAMOS EL ESTÁNDAR DE LA INDUSTRIA: 224x224
transform = transforms.Compose([
    transforms.Resize((224, 224)), # <--- CALIDAD MÁXIMA
    transforms.RandomHorizontalFlip(),
    transforms.RandomCrop(224, padding=16), # Un poco más de padding por ser más grande
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

full_dataset = torchvision.datasets.Flowers102(root='./data', split='train', download=True, transform=transform)
num_classes = 102

train_size = int(0.8 * len(full_dataset))
val_size = len(full_dataset) - train_size
train_dataset, val_dataset = random_split(full_dataset, [train_size, val_size])

# BAJAMOS EL BATCH SIZE A 16 PARA QUE NO EXPLOTE LA MEMORIA DE COLAB
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=16, shuffle=False)

# ══════════════════════════════════════════════════════════════════
# TU FUNCIÓN GENAL
# ══════════════════════════════════════════════════════════════════
class GenalShift(nn.Module):
    def __init__(self, theta_init: float = 1.0, beta_init: float = 0.0):
        super().__init__()
        self.theta = nn.Parameter(torch.tensor(theta_init, dtype=torch.float32))
        self.beta = nn.Parameter(torch.tensor(beta_init, dtype=torch.float32))
    def forward(self, x):
        k = torch.clamp(torch.nn.functional.softplus(self.theta), min=1e-6, max=2.0)
        return x * torch.sigmoid((x - self.beta) / k)

# ══════════════════════════════════════════════════════════════════
# EL MODELO (CNN)
# ══════════════════════════════════════════════════════════════════
class BotanicoCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
        self.act1 = GenalShift()
        self.pool = nn.MaxPool2d(2, 2)

        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.act2 = GenalShift()

        # AHORA LA IMAGEN LLEGA A 56x56 DESPUÉS DE LOS DOS POOLINGS
        self.fc1 = nn.Linear(32 * 56 * 56, 128)
        self.act3 = GenalShift()
        self.drop = nn.Dropout(0.5)
        self.fc2 = nn.Linear(128, num_classes)

    def forward(self, x):
        x = self.pool(self.act1(self.conv1(x)))
        x = self.pool(self.act2(self.conv2(x)))
        x = x.view(-1, 32 * 56 * 56) # <--- APLANADO ACTUALIZADO
        x = self.drop(self.act3(self.fc1(x)))
        x = self.fc2(x)
        return x

# ══════════════════════════════════════════════════════════════════
# ENTRENAMIENTO
# ══════════════════════════════════════════════════════════════════
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
modelo = BotanicoCNN(num_classes).to(device)
criterio = nn.CrossEntropyLoss()
optimizer = optim.Adam(modelo.parameters(), lr=0.001)

print("🚀 Iniciando entrenamiento 224x224 con GenalShift...")
epocas = 15 # Subimos un poquito las épocas
historial_loss = []

for epoch in range(epocas):
    modelo.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = modelo(images)
        loss = criterio(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    avg_loss = running_loss / len(train_loader)
    historial_loss.append(avg_loss)
    print(f"Época {epoch+1}/{epocas} | Pérdida: {avg_loss:.4f}")

# Validación
modelo.eval()
correct = 0
total = 0
with torch.no_grad():
    for images, labels in val_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = modelo(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

precision = 100 * correct / total
print(f"\n🏆 ¡Entrenamiento terminado!")
print(f"🎯 Precisión (Accuracy): {precision:.2f}%")

plt.figure(figsize=(8, 4))
plt.plot(historial_loss, marker='o', color='blue', linewidth=2)
plt.title("Aprendizaje 224x224 (usando GenalShift)", fontweight='bold')
plt.xlabel("Épocas")
plt.ylabel("Pérdida (Loss)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader, random_split
import matplotlib.pyplot as plt
import time

# 1. Preparar Datos (Usamos 64x64 para que sea rápido en el celular)
print("📥 Preparando datos...")
transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomCrop(64, padding=4),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

full_dataset = torchvision.datasets.Flowers102(root='./data', split='train', download=True, transform=transform)
num_classes = 102
train_size = int(0.8 * len(full_dataset))
val_size = len(full_dataset) - train_size
train_dataset, val_dataset = random_split(full_dataset, [train_size, val_size])
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)

# ══════════════════════════════════════════════════════════════════
# LOS CONTENDIENTES (Funciones de Activación)
# ══════════════════════════════════════════════════════════════════
class GenalShift(nn.Module):
    def __init__(self, theta_init=1.0, beta_init=0.0):
        super().__init__()
        self.theta = nn.Parameter(torch.tensor(theta_init))
        self.beta = nn.Parameter(torch.tensor(beta_init))
    def forward(self, x):
        k = torch.clamp(F.softplus(self.theta), min=1e-6, max=2.0)
        return x * torch.sigmoid((x - self.beta) / k)

class Swish(nn.Module):
    def forward(self, x):
        return x * torch.sigmoid(x)

# ══════════════════════════════════════════════════════════════════
# LA ARQUITECTURA (3 Capas para más poder)
# ══════════════════════════════════════════════════════════════════
class CNN_Model(nn.Module):
    def __init__(self, activation_class, num_classes=102):
        super().__init__()
        self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
        self.act1 = activation_class()
        self.pool = nn.MaxPool2d(2, 2)

        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.act2 = activation_class()

        self.conv3 = nn.Conv2d(32, 64, 3, padding=1)
        self.act3 = activation_class()

        # 64x64 -> pool -> 32 -> pool -> 16 -> pool -> 8
        self.fc1 = nn.Linear(64 * 8 * 8, 256)
        self.act4 = activation_class()
        self.drop = nn.Dropout(0.5)
        self.fc2 = nn.Linear(256, num_classes)

    def forward(self, x):
        x = self.pool(self.act1(self.conv1(x)))
        x = self.pool(self.act2(self.conv2(x)))
        x = self.pool(self.act3(self.conv3(x)))
        x = x.view(-1, 64 * 8 * 8)
        x = self.drop(self.act4(self.fc1(x)))
        x = self.fc2(x)
        return x

# ══════════════════════════════════════════════════════════════════
# FUNCIÓN DE ENTRENAMIENTO
# ══════════════════════════════════════════════════════════════════
def entrenar_modelo(act_class, nombre, epocas=15):
    print(f"\n⚔️  Iniciando entrenamiento de: {nombre}...")
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    modelo = CNN_Model(act_class).to(device)
    criterio = nn.CrossEntropyLoss()
    optimizer = optim.Adam(modelo.parameters(), lr=0.001)

    historial_loss = []
    t0 = time.time()

    for epoch in range(epocas):
        modelo.train()
        running_loss = 0.0
        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = modelo(images)
            loss = criterio(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()

        avg_loss = running_loss / len(train_loader)
        historial_loss.append(avg_loss)
        print(f"   Época {epoch+1}/{epocas} | Pérdida: {avg_loss:.4f}")

    # Validación final
    modelo.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = modelo(images)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    precision = 100 * correct / total
    tiempo_total = time.time() - t0
    print(f"🏁 {nombre} terminó. Precisión: {precision:.2f}% en {tiempo_total/60:.1f} min")

    return historial_loss, precision

# ══════════════════════════════════════════════════════════════════
# LA BATALLA FINAL
# ══════════════════════════════════════════════════════════════════
resultados = {}

# 1. ReLU
loss_relu, acc_relu = entrenar_modelo(nn.ReLU, "ReLU")
resultados['ReLU'] = {'loss': loss_relu, 'acc': acc_relu}

# 2. GELU
loss_gelu, acc_gelu = entrenar_modelo(nn.GELU, "GELU")
resultados['GELU'] = {'loss': loss_gelu, 'acc': acc_gelu}

# 3. Swish
loss_swish, acc_swish = entrenar_modelo(Swish, "Swish")
resultados['Swish'] = {'loss': loss_swish, 'acc': acc_swish}

# 4. GenalShift (Tu tecnología)
loss_genal, acc_genal = entrenar_modelo(GenalShift, "GenalShift")
resultados['GenalShift'] = {'loss': loss_genal, 'acc': acc_genal}

# ══════════════════════════════════════════════════════════════════
# RESULTADOS FINALES
# ══════════════════════════════════════════════════════════════════
print("\n" + "="*50)
print(" 🏆 TABLA DE CLASIFICACIÓN FINAL (FLOWERS-102) 🏆")
print("="*50)
for nombre, data in sorted(resultados.items(), key=lambda x: x[1]['acc'], reverse=True):
    print(f" {nombre:<12} | Precisión: {data['acc']:.2f}%")

# Gráfica de las 4 curvas
plt.figure(figsize=(10, 6))
plt.plot(resultados['ReLU']['loss'], label=f"ReLU ({resultados['ReLU']['acc']:.1f}%)", color='red', linestyle='--')
plt.plot(resultados['GELU']['loss'], label=f"GELU ({resultados['GELU']['acc']:.1f}%)", color='green', linestyle='-.')
plt.plot(resultados['Swish']['loss'], label=f"Swish ({resultados['Swish']['acc']:.1f}%)", color='orange', linestyle=':')
plt.plot(resultados['GenalShift']['loss'], label=f"GenalShift ({resultados['GenalShift']['acc']:.1f}%)", color='blue', linewidth=3)

plt.title("La Batalla de las Activaciones (Flowers-102)", fontweight='bold', fontsize=14)
plt.xlabel("Épocas", fontsize=12)
plt.ylabel("Pérdida (Loss)", fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
