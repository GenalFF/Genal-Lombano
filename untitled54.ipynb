```python
# 1. Instalamos tu librería y las herramientas de visión
!pip install genal-activation torch torchvision matplotlib

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
from genal_activation import GenalShift # <--- Tu tecnología

# 2. Preparamos las imágenes (las convertimos a tensores y las normalizamos)
transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.5,), (0.5,))])

# Descargamos los datos de entrenamiento y prueba
trainset = torchvision.datasets.FashionMNIST(root='./data', train=True, download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=64, shuffle=True)

testset = torchvision.datasets.FashionMNIST(root='./data', train=False, download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=64, shuffle=False)

# 3. Definimos la Red Neuronal (CNN) con Genal Activation
class VisionGenalNet(nn.Module):
    def __init__(self):
        super(VisionGenalNet, self).__init__()
        # Capas Convolucionales (Para detectar bordes y formas)
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, padding=1)
        self.act1 = GenalShift() # <--- Inyectamos GenalShift aquí
        self.pool = nn.MaxPool2d(2, 2)

        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.act2 = GenalShift() # <--- Y aquí

        # Capas Lineales (Para clasificar)
        self.fc1 = nn.Linear(64 * 7 * 7, 128)
        self.act3 = GenalShift() # <--- Y aquí
        self.fc2 = nn.Linear(128, 10) # 10 categorías de ropa

    def forward(self, x):
        x = self.pool(self.act1(self.conv1(x)))
        x = self.pool(self.act2(self.conv2(x)))
        x = x.view(-1, 64 * 7 * 7) # Aplanar la imagen
        x = self.act3(self.fc1(x))
        x = self.fc2(x)
        return x

# 4. Inicializamos el modelo, la función de pérdida y el optimizador
modelo = VisionGenalNet()
criterio = nn.CrossEntropyLoss()
optimizador = optim.Adam(modelo.parameters(), lr=0.001)

# 5. ¡A ENTRENAR! (Solo 3 épocas para que sea rápido en Colab)
print("Iniciando entrenamiento con GenalShift...")
for epoch in range(3):
    perdida_total = 0.0
    for i, datos in enumerate(trainloader, 0):
        entradas, etiquetas = datos
        optimizador.zero_grad()

        salidas = modelo(entradas)
        perdida = criterio(salidas, etiquetas)
        perdida.backward()
        optimizador.step()

        perdida_total += perdida.item()

    print(f"Época {epoch + 1} completada - Pérdida: {perdida_total / len(trainloader):.4f}")

print("¡Entrenamiento terminado!")

# 6. Probar el modelo con una imagen real
def mostrar_prediccion():
    clases = ['Camiseta', 'Pantalón', 'Suéter', 'Vestido', 'Abrigo',
              'Sandalia', 'Camisa', 'Zapatilla', 'Bolso', 'Bota']

    dataiter = iter(testloader)
    imagenes, etiquetas = next(dataiter)

    # Predecir la primera imagen
    with torch.no_grad():
        salida = modelo(imagenes[0].unsqueeze(0))
        _, prediccion = torch.max(salida, 1)

    # Mostrar la imagen
    plt.imshow(imagenes[0].squeeze(), cmap='gray')
    plt.title(f"El modelo dice que es: {clases[prediccion.item()]}\nRealmente es: {clases[etiquetas[0].item()]}")
    plt.show()

mostrar_prediccion()

# 7. Ver los parámetros que GenalShift aprendió por sí solo
k1, beta1 = modelo.act1.get_k(), modelo.act1.get_beta()
print(f"⚙️ Parámetros GenalShift aprendidos en Capa 1: k={k1:.4f}, β={beta1:.4f}")
```
